## Task 1: GroupBy (3 min)

Total revenue (`amount`) and number of orders per `customer_id`.
Sort by total revenue descending. Show top 10.

````python
# YOUR CODE HERE
customers.merge(orders, on="customer_id", how="left").groupby("customer_id")["amount"].agg("sum", "count").sort_values(ascending=False).head(10)
````

## Task 2: Merge (3 min)

Merge `orders` with `customers` on `customer_id`.
Calculate **average order value** (`amount`) by `plan_type`.

````python
# YOUR CODE HERE
customers["plan_type"] = customers["plan_type"].str.title()
customers.merge(orders, on="customer_id", how="left").groupby("plan_type")["amount"].mean()
````
